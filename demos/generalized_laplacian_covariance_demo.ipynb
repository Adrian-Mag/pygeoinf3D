# Gaussian Random Fields from Negative Laplacian Powers

This notebook demonstrates **Gaussian random fields** whose covariance operator is an integer power of the positive box Laplacian:

$$C = (-\alpha\,\Delta)^s, \quad s \in \{-1,\,-2,\,-3\}$$

on the unit box $[0,1]^3$ with Dirichlet boundary conditions.

**Key ideas:**
- The operator $(-\alpha\Delta)^s$ is implemented by `GeneralizedBoxLaplacian` with `power=s`.
- For $s < 0$ the covariance eigenvalues $\mu_k = (\alpha\lambda_k)^s$ *decay* with mode index — low-frequency modes carry the most variance.
- More negative $s$ means faster eigenvalue decay → smoother fields.
- `KLSampler3D` accepts any `SpectralOperator3D` as a covariance, so all three powers plug in without changes.
