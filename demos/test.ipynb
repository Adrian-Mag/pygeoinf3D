```python
from pygeoinf3d import BoxDomain, Lebesgue3D, Path3D
from pygeoinf3d import plot
import numpy as np
```

```python
domain = BoxDomain(np.array([0, 0, 0]), np.array([1, 1, 1]))
M = Lebesgue3D(domain)
path = Path3D.from_points(np.array([[0, 0, 0], [1, 1, 1]]))
```

```python
plot.plot_paths3d_interactive(path, show_plot=False)
```

```python
from pygeoinf3d import BesselSobolevBox, BoxLaplacianSpectrumProvider, Sobolev3D
provider = BoxLaplacianSpectrumProvider(M, truncation=(10,10,10))
A = BesselSobolevBox(M, provider, shift=100, order=2.0)
A_inv = BesselSobolevBox(M, provider, shift=100, order=-2.0)
M_sobolev = Sobolev3D(M, A, A_inv)
```

```python
from pygeoinf3d import PathIntegralFunctional

l = PathIntegralFunctional(M_sobolev, path)
representer = M_sobolev.from_dual(l)
```

```python
plot.plot_function3d_axis_slider(representer, show_plot=False, n_slices=10, plot_n_modes=1000)
```

```python
plot.plot_function3d_isosurface(representer, show_plot=False)
```
